```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_california_housing
```

```python
housing = fetch_california_housing()

df = pd.DataFrame(housing.data, columns=housing.feature_names)
df['MedHouseVal'] = housing.target
```

```python
print(df.info())
print(df.isnull().sum())
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
standard_scaler = StandardScaler()
standard_scaled_data=standard_scaler.fit_transform(df)
standard_scaled_data
```

2. Regression Algorithm Implementation:
● Implement the following regression algorithms:
○ Linear Regression
○ Decision Tree Regressor
○ Random Forest Regressor
○ Gradient Boosting Regressor
○ Support Vector Regressor (SVR)
● For each algorithm: Provide a brief explanation of how it works. Explain why it
might be suitable for this dataset.

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR
```

```python
from sklearn.model_selection import train_test_split
```

```python
lnr=LinearRegression()
x=df.drop('MedHouseVal',axis=1)
y=df['MedHouseVal']
X_train, X_test, y_train, y_test=train_test_split(x, y, random_state=42)
lnr.fit(X_train,y_train)
print(lnr.score(X_train,y_train))
```

```python
rf = RandomForestRegressor()
rf.fit(X_train, y_train)
print(rf.score(X_train,y_train))
```

```python
dt=DecisionTreeRegressor()
dt.fit(X_train, y_train)
print(dt.score(X_train,y_train))
```

```python
gdb=GradientBoostingRegressor()
gdb.fit(X_train, y_train)
print(gdb.score(X_train,y_train))
```

```python
svcr=SVR()
svcr.fit(X_train, y_train)
print(svcr.score(X_train,y_train))
```

3. Model Evaluation and Comparison:
● Evaluate the performance of each algorithm using the following metrics:
○ Mean Squared Error (MSE)
○ Mean Absolute Error (MAE)
○ R-squared Score (R2)
● Compare the results of all models and identify: The best-performing algorithm
with justification and the worst-performing algorithm with reasoning.

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error , r2_score
```

```python
predictions = lnr.predict(X_test)
print("Mean Squared Error:", mean_squared_error(y_test, predictions))
print("Mean Absolute Error:", mean_absolute_error(y_test, predictions))
print("R-squared Score:", r2_score(y_test, predictions))
```

```python
predictions = rf.predict(X_test)
print("Mean Squared Error:", mean_squared_error(y_test, predictions))
print("Mean Absolute Error:", mean_absolute_error(y_test, predictions))
print("R-squared Score:", r2_score(y_test, predictions))
```

```python
predictions = dt.predict(X_test)
print("Mean Squared Error:", mean_squared_error(y_test, predictions))
print("Mean Absolute Error:", mean_absolute_error(y_test, predictions))
print("R-squared Score:", r2_score(y_test, predictions))
```

```python
predictions = gdb.predict(X_test)
print("Mean Squared Error:", mean_squared_error(y_test, predictions))
print("Mean Absolute Error:", mean_absolute_error(y_test, predictions))
print("R-squared Score:", r2_score(y_test, predictions))
```

```python
predictions = svcr.predict(X_test)
print("Mean Squared Error:", mean_squared_error(y_test, predictions))
print("Mean Absolute Error:", mean_absolute_error(y_test, predictions))
print("R-squared Score:", r2_score(y_test, predictions))
```

### Analysis:

**Best Performing Algorithm: Random Forest Regressor**

The Random Forest Regressor stands out as the best-performing algorithm. It has the **lowest Mean Squared Error (0.254)** and **Mean Absolute Error (0.329)**, indicating that its predictions are closest to the actual values on average. Crucially, it also boasts the **highest R-squared score (0.808)**. An R-squared value of 0.808 means that approximately 80.8% of the variance in the target variable (Median House Value) can be explained by the independent variables in the model. This high R-squared, coupled with low errors, suggests a robust model that fits the data very well and generalizes effectively to unseen data.

**Worst Performing Algorithm: Support Vector Regressor (SVR)**

The Support Vector Regressor (SVR) is the worst-performing algorithm in this comparison. It exhibits the **highest Mean Squared Error (1.353)** and **Mean Absolute Error (0.867)**, meaning its predictions have the largest deviations from the actual values. More notably, its **R-squared score is -0.022**. A negative R-squared score indicates that the model performs worse than a simple horizontal line (the mean of the target variable). This suggests that the SVR model, in its current configuration, is not capturing any meaningful patterns in the data and is effectively a very poor fit for this dataset.
